# Etapa 1 — Recinto y población inicial

**Objetivo:** visualizar el recinto y verificar la población inicial. Esta notebook funciona sin completar humo, BFS ni movimiento.

**Qué se prueba:** que se generen exactamente 50 agentes, todos en posiciones diferentes, con 25 de activación alta y 25 de baja, sin personas sobre paredes ni sobre el foco del incendio.

```python
import numpy as np
import matplotlib.pyplot as plt
from modelo import Configuracion, Entorno, crear_agentes_iniciales
from visualizacion import mostrar_escenario

# Crear el recinto y distribuir 50 agentes con una semilla fija.
configuracion = Configuracion(n_agentes=50, semilla=42)
entorno = Entorno(configuracion)
agentes = crear_agentes_iniciales(configuracion, entorno, np.random.default_rng(configuracion.semilla))

# Mostrar el escenario sin necesidad de calcular BFS ni propagar humo.
figura = mostrar_escenario(entorno, agentes)
plt.show()
```

```python
# Debe existir exactamente la cantidad de personas solicitada.
assert len(agentes) == 50

# Un conjunto elimina posiciones repetidas: si contiene 50 posiciones,
# sabemos que no hay dos agentes en una misma celda.
assert len({a.posicion() for a in agentes}) == 50

# La mitad de los agentes debe tener activación alta y la otra mitad baja.
assert sum(a.tipo == "alta" for a in agentes) == 25
assert sum(a.tipo == "baja" for a in agentes) == 25

# Ningún agente debe comenzar sobre una pared o el foco de incendio.
assert all(not entorno.paredes[a.y, a.x] for a in agentes)
assert all(a.posicion() not in entorno.fuentes for a in agentes)

print("Etapa 1: verificaciones aprobadas.")
```
